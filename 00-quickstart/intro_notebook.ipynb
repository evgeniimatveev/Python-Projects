Welcome to the **intro Jupyter notebook** for the Python Projects repository.

In this notebook you will:
- See a quick overview of the numbered project folders
- Load them into a small Pandas DataFrame
- Sort projects by difficulty level

Use this as a starting point before exploring the individual project folders.


```python
# Quick overview of projects using Pandas

import pandas as pd

projects = [
    ("01-avocado-price-forecasting", "Avocado Price Forecasting", "ML / Time Series", "intermediate"),
    ("02-amazon-reviews-analysis", "Amazon Reviews Sentiment", "NLP", "intermediate"),
    ("03-real-time-ai-detection", "Real-Time AI Filters", "Computer Vision", "advanced"),
    ("04-movie-recommendation", "Movie Recommendation App", "Recommender System", "intermediate"),
    ("05-image-processing-app", "Image Processing App", "OpenCV", "intermediate"),
    ("06-opencv-realtime-effects", "OpenCV Real-time Processing", "Computer Vision", "advanced"),
    ("07-tic-tac-toe", "Tic Tac Toe with minimax AI", "Game / AI", "intermediate"),
    ("08-fortune-teller", "Fortune Teller Game", "Game / Logic", "beginner"),
    ("09-guessing-game", "Number Guessing Game", "Game", "beginner"),
    ("10-zodiac-sign-finder", "Zodiac Sign Finder", "Utility", "beginner"),
]

df = pd.DataFrame(projects, columns=["folder", "title", "topic", "level"])
df
```

```python
# Sort projects by difficulty level

level_order = ["beginner", "intermediate", "advanced"]
df["level"] = pd.Categorical(df["level"], categories=level_order, ordered=True)

df_sorted = df.sort_values("level")
df_sorted
```
