## 1. Core Calculations — Zodiac & Life-Path Number

```python
import datetime
import random
import itertools

ZODIAC_DATES = [
    ("Capricorn",   12, 22,  1, 19), ("Aquarius",    1, 20,  2, 18),
    ("Pisces",       2, 19,  3, 20), ("Aries",       3, 21,  4, 19),
    ("Taurus",       4, 20,  5, 20), ("Gemini",      5, 21,  6, 20),
    ("Cancer",       6, 21,  7, 22), ("Leo",         7, 23,  8, 22),
    ("Virgo",        8, 23,  9, 22), ("Libra",       9, 23, 10, 22),
    ("Scorpio",     10, 23, 11, 21), ("Sagittarius", 11, 22, 12, 21),
    ("Capricorn",   12, 22,  1, 19),
]

def get_zodiac_sign(day, month):
    for sign, sm, sd, em, ed in ZODIAC_DATES:
        if (month == sm and day >= sd) or (month == em and day <= ed):
            return sign
    return "Unknown"

def life_path_number(year):
    """Reduce year to single digit (numerology life-path)."""
    while year >= 10:
        year = sum(int(d) for d in str(year))
    return year
```

## 2. Lucky Extras — Element, Color, Numbers & Message

```python
ELEMENTS = {
    "Fire":  ["Aries","Leo","Sagittarius"],
    "Earth": ["Taurus","Virgo","Capricorn"],
    "Air":   ["Gemini","Libra","Aquarius"],
    "Water": ["Cancer","Scorpio","Pisces"],
}

LUCKY_COLORS = {
    "Aries":"Red ❤️","Taurus":"Green 💚","Gemini":"Yellow 💛","Cancer":"White 🤍",
    "Leo":"Gold ✨","Virgo":"Brown 🤎","Libra":"Pink 🎀","Scorpio":"Black 🖤",
    "Sagittarius":"Purple 💜","Capricorn":"Grey ⚫","Aquarius":"Blue 💙","Pisces":"Sea Green 🌊",
}

MYSTICAL_MESSAGES = [
    "✨ The universe is aligning in your favour...",
    "🌙 Trust your intuition — it will guide you.",
    "🔮 A significant change is coming; embrace it.",
    "🔥 Your inner fire is stronger than you think.",
    "🌿 A moment of calm will bring unexpected clarity.",
]

def get_element(sign):
    for element, signs in ELEMENTS.items():
        if sign in signs: return element
    return "Unknown"

def get_lucky_numbers(name):
    random.seed(sum(ord(c) for c in name))
    return random.sample(range(1, 100), 3)

def build_report(name, day, month, year):
    sign    = get_zodiac_sign(day, month)
    num     = life_path_number(year)
    element = get_element(sign)
    color   = LUCKY_COLORS.get(sign, "Unknown")
    lucky   = get_lucky_numbers(name)
    msg     = random.choice(MYSTICAL_MESSAGES)
    age     = datetime.date.today().year - year
    return (
        f"\n━━━━━━━━━━━━━━━━━━━━━━\n"
        f"🔮  YOUR ASTRO REPORT\n"
        f"━━━━━━━━━━━━━━━━━━━━━━\n"
        f"👤  Name:             {name}\n"
        f"🎂  Birthdate:        {day:02d}/{month:02d}/{year}  (Age: {age})\n"
        f"✨  Zodiac Sign:      {sign}\n"
        f"🔢  Life-Path Number: {num}\n"
        f"🔥  Lucky Element:    {element}\n"
        f"🎨  Lucky Color:      {color}\n"
        f"🎲  Lucky Numbers:    {', '.join(map(str, lucky))}\n"
        f"🌠  Message:          {msg}\n"
        f"━━━━━━━━━━━━━━━━━━━━━━\n"
        f"May the stars guide your path! ✨\n"
    )
```

## 3. Pure Python Demo — No GUI Required

```python
# Run this cell to test the logic directly in the notebook
print(build_report("Evgenii", 15, 3, 1990))
print(build_report("Alice",   1,  1, 2000))
```

## 4. Full GUI Application (tkinter)

Run the cell below to launch the desktop app. Requires a display environment (local Jupyter, VS Code, etc.).

```python
import tkinter as tk
from tkinter import messagebox

try:
    from reportlab.pdfgen import canvas as pdf_canvas
    PDF_AVAILABLE = True
except ImportError:
    PDF_AVAILABLE = False
    print("⚠️  reportlab not installed — PDF export disabled. Run: pip install reportlab")

def animate_bg(root):
    colors = itertools.cycle(["#222831","#2a3137","#32383e","#3a4046","#444b52"])
    def cycle(): root.config(bg=next(colors)); root.after(1200, cycle)
    cycle()

def save_pdf(text):
    if not PDF_AVAILABLE:
        messagebox.showwarning("PDF","Install reportlab: pip install reportlab"); return
    c = pdf_canvas.Canvas("Astro_Report.pdf")
    c.setFont("Helvetica", 11)
    y = 800
    for line in text.split("\n"):
        c.drawString(50, y, line); y -= 18
        if y < 50: c.showPage(); y = 800
    c.save()
    messagebox.showinfo("Saved","Report saved as Astro_Report.pdf ✅")

def generate():
    name = name_entry.get().strip()
    try: day, month, year = int(day_e.get()), int(month_e.get()), int(year_e.get())
    except ValueError: messagebox.showerror("Error","Enter valid numbers."); return
    if not name: messagebox.showerror("Error","Please enter your name."); return
    result_label.config(text=build_report(name, day, month, year))

root = tk.Tk()
root.title("Zodiac & Numerology Analyzer")
root.geometry("560x720")
animate_bg(root)

tk.Label(root,text="🔮 Astro Report Generator",font=("Arial",18,"bold"),fg="#FFD700",bg="#222831").pack(pady=12)

def lbl(t): tk.Label(root,text=t,font=("Arial",11),fg="white",bg="#222831").pack()
def entry(): e=tk.Entry(root,font=("Arial",12),width=26,justify="center"); e.pack(); return e

lbl("Name:");   name_entry = entry()
lbl("Day:");    day_e      = entry()
lbl("Month:");  month_e    = entry()
lbl("Year:");   year_e     = entry()

tk.Button(root,text="✨ Generate",command=generate,bg="#00cc99",fg="white",font=("Arial",13,"bold")).pack(pady=8)
tk.Button(root,text="💾 Save PDF",command=lambda:save_pdf(result_label.cget("text")),bg="#ff9966",fg="white",font=("Arial",13,"bold")).pack(pady=4)

result_label = tk.Label(root,text="",wraplength=480,font=("Arial",11),justify="left",fg="white",bg="#393E46")
result_label.pack(pady=10)

root.mainloop()
```
