## Overview

Recommends a movie based on your age, covering every age group from toddlers to seniors.

**How to play:** Run the last cell and enter your age.

**Dependencies:** standard library only.

## 1. Movie Database

```python
MOVIES = {
    (0,  3):  ("Baby Shark's Big Show 🦈",          "Engaging musical adventures for toddlers."),
    (3,  6):  ("Peppa Pig 🐷",                      "Lighthearted family stories loved by preschoolers."),
    (6,  9):  ("Toy Story 🎠",                      "Friendship and adventure as toys come to life."),
    (9,  12): ("Inside Out 🎭",                     "A touching exploration of emotions and growing up."),
    (12, 15): ("Spider-Man: Into the Spider-Verse 🕸️","Visually stunning superhero story about courage."),
    (15, 18): ("Black Panther 🐾",                  "Cultural phenomenon with powerful storytelling."),
    (18, 21): ("Joker 🃏",                          "Deep psychological drama — birth of an icon."),
    (21, 24): ("Parasite 🎞️",                      "Genre-bending masterpiece of social commentary."),
    (24, 27): ("Interstellar 🌌",                   "Sci-fi epic exploring love, time, and survival."),
    (27, 30): ("The Matrix 🟩",                     "Groundbreaking thriller about the nature of reality."),
    (30, 40): ("The Godfather 👑",                  "Classic mafia drama — power, family, betrayal."),
    (40, 50): ("Schindler's List 📜",               "A deeply moving drama about heroism in darkness."),
    (50, 60): ("Forrest Gump 🏃",                   "An extraordinary man's journey through history."),
    (60, 70): ("The Sound of Music 🎼",              "A timeless musical about love and resilience."),
    (70, 81): ("Casablanca 🕊️",                     "A romantic classic with unforgettable performances."),
    (81, 200):("It's a Wonderful Life 🎄",         "A feel-good classic about life's true value."),
}

TRENDING = [
    "Avatar: The Way of Water 🌊",
    "Top Gun: Maverick ✈️",
    "Everything Everywhere All At Once 🌌",
    "Dune: Part Two 🏜️",
    "Oppenheimer ☢️",
]
```

## 2. Recommendation Logic

```python
def recommend_movie(age):
    """Return (title, summary) for the given age."""
    if age < 0:
        return None
    for (lo, hi), (title, summary) in MOVIES.items():
        if lo <= age < hi:
            return title, summary
    return None

def show_trending():
    print("\n🔥 Trending right now:")
    for m in TRENDING:
        print(f"   🎥 {m}")
    print()
```

## 3. App Loop

```python
def movie_app():
    print("🎬 Movie Recommendation App")
    print("Enter your age and get a personalised pick. Type -1 to quit.\n")
    show_trending()

    while True:
        try:
            age = int(input("Your age (or -1 to quit): ").strip())
            if age == -1:
                print("\nThanks for using the app. Enjoy your movie! 🍿")
                break
            result = recommend_movie(age)
            if result:
                title, summary = result
                print(f"\n🎥  {title}")
                print(f"📝  {summary}\n")
            else:
                print("❌ Please enter a valid age (0 or above).\n")
        except ValueError:
            print("❌ Enter a whole number.\n")
```

## ▶ Run the App

```python
movie_app()
```
