## Overview

A number guessing game where the AI adapts its hints based on how close you are.

**How to play:** Run the last cell — choose a difficulty, then guess the number.

**Dependencies:** `random`, `time`, `sys` — all standard library, no installs needed.

## 1. Setup — Imports, Colors & Messages

```python
import random
import time
import sys

def colored(text, color_code):
    """Apply ANSI color to text (works in real terminals, stripped in Jupyter)."""
    return f"\033[{color_code}m{text}\033[0m" if sys.stdout.isatty() else text

SUCCESS_MESSAGES = [
    "🎉 You're a legend! 🏆",
    "🔥 Incredible! You nailed it! 🎯",
    "🎯 Boom! Headshot! 💥",
    "🚀 Unstoppable! You are on fire! 🔥",
    "🧠 Mastermind detected! Genius mode activated! 🚀",
]

ASCII_ARTS = [
    "/\_/\\\n ( -.- )  🎉 You did it!\n  (>💖<)\n\n🎊🎉 YOU ARE A WINNER! 🎉🎊",
    "/\_/\\\n ( ^_^ )  🏅 Champion!\n  (>🐾<)\n\n🎊🎉 YOU ARE A WINNER! 🎉🎊",
    "/\_/\\\n ( o.o )  🎉 Meow! You won!\n  > ^ <\n\n🎊🎉 YOU ARE A WINNER! 🎉🎊",
]
```

## 2. AI Hint Engine

```python
def ai_hint(guess, true_number, low, high):
    """Return an adaptive hint scaled to how far off the guess is."""
    diff = abs(guess - true_number)
    rng  = high - low
    if diff > rng // 2:
        direction = "lower" if guess > true_number else "higher"
        return f"🔥 Way off! Go {diff // 2}+ numbers {direction}!"
    elif diff > rng // 4:
        lo, hi = min(guess, true_number), max(guess, true_number)
        return f"🎯 Closer — the answer is between {lo} and {hi}!"
    elif diff > rng // 10:
        return f"🧠 Very close — within {diff} numbers!"
    elif diff <= 3:
        return "🎯 Super close! Within 3 numbers!"
    else:
        return "🤔 Keep going — you can do it!"
```

## 3. Game Logic

```python
def guessing_game():
    """Main game loop: choose difficulty, guess, get hints, win."""
    print(colored("\n🎯 Welcome to Guess the Number!\n", "36"))
    levels = {
        "1": (1, 10,  "🐣 Easy   (1–10)"),
        "2": (1, 49,  "⚡ Medium (1–49)"),
        "3": (1, 100, "🔥 Hard   (1–100)"),
    }
    while True:
        print("Select difficulty:\n1️⃣  Easy (1-10)\n2️⃣  Medium (1-49)\n3️⃣  Hard (1-100)")
        choice = input("Enter 1, 2, or 3: ").strip()
        if choice in levels:
            low, high, name = levels[choice]
            print(colored(f"\n🎮 {name} — let's go!\n", "34"))
            break
        print(colored("❌ Enter 1, 2 or 3.\n", "31"))

    secret   = random.randint(low, high)
    attempts = 0
    start    = time.time()

    while True:
        try:
            raw = input(colored(f"💡 Guess ({low}–{high}) or 'exit': ", "33")).strip()
            if raw.lower() == "exit":
                print(colored("👋 Thanks for playing!", "36"))
                return
            guess = int(raw)
            if not (low <= guess <= high):
                print(colored(f"⚠️  Enter a number between {low} and {high}.\n", "33"))
                continue
            attempts += 1
            print(colored("🤔 Checking...", "35"))
            time.sleep(0.4)
            if guess == secret:
                elapsed = round(time.time() - start, 2)
                msg = random.choice(SUCCESS_MESSAGES)
                print(colored(f"\n🎉 CORRECT! You got {secret} in {attempts} attempt(s) and {elapsed}s!", "32"))
                print(colored(f"🏆 {msg}\n", "33"))
                print(colored(random.choice(ASCII_ARTS), "36"))
                return
            print(colored(ai_hint(guess, secret, low, high) + "\n", "36"))
        except ValueError:
            print(colored("⚠️  Please enter a valid number.\n", "33"))
```

## ▶ Run the Game

```python
guessing_game()
```
