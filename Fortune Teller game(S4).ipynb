## Overview

An interactive fortune telling game. Choose a color, pick your sacred number — the Oracle reveals your destiny.

**How to play:** Run the last cell and follow the prompts.

**Dependencies:** `random`, `time` — standard library only.

## 1. Setup — Predictions & Meanings

```python
import random
import time

PREDICTIONS = {
    "yellow": {3, 12, 21, 33, 48},
    "green":  {7, 16, 22, 37, 45},
    "blue":   {5, 11, 19, 28, 42},
    "red":    {9, 14, 27, 35, 50},
}

NUMBER_MEANINGS = {
    3:"🎭 An unexpected event will change everything!",
    5:"🌀 A path you never considered will open...",
    7:"💰 An opportunity for wealth is near!",
    9:"🦁 Your strength will be tested soon...",
    11:"🔑 A long-kept secret will come to light...",
    12:"📖 A powerful lesson will shape your future...",
    14:"🔥 Passion and energy will fuel your next move!",
    16:"✨ Your destiny is shifting — pay attention!",
    19:"🌙 A powerful transformation awaits...",
    21:"🕊 A sign from the universe will reveal itself...",
    22:"🤝 A fateful meeting will change your course!",
    27:"🔄 The past is not finished with you yet...",
    28:"🌍 Your journey leads to new horizons...",
    33:"🌟 A rare blessing is about to enter your life!",
    35:"⚔️ A great challenge will prove your strength!",
    37:"💭 Your dreams hold a message — listen closely!",
    42:"💫 The answer you seek is already within you...",
    45:"📜 A prophecy written long ago concerns you...",
    48:"🔮 Trust your intuition — it is rarely wrong...",
    50:"🌀 The cycle is ending, prepare for renewal...",
}

RARE_FORTUNES = [
    "🌌✨ The universe has chosen you for a legendary destiny! 🚀",
    "💎🏆 You will leave a legacy remembered for generations!",
    "⚡🌠 A great power within you is awakening!",
]

history = []
```

## 2. Helper Functions — Animation & Input

```python
def mystical_loading():
    messages = [
        "✨ The energies are shifting...",
        "🔮 The Oracle is focusing...",
        "🌌 The stars are aligning...",
        "🕯️  The spirits are whispering...",
        "🌠 A great vision is forming...",
    ]
    for msg in messages:
        print(f"\r{msg}    ", end="", flush=True)
        time.sleep(random.uniform(0.6, 1.0))
    print("\n✨ The vision is clear! ✨\n")

def reveal_numbers(color):
    print("\n📜 The spirits whisper sacred numbers to you...")
    time.sleep(0.6)
    for num in sorted(PREDICTIONS[color]):
        print(f"  ✨ {num:>2} ✨  —  {NUMBER_MEANINGS[num]}")
        time.sleep(0.2)
    print()

def ask(prompt, options):
    while True:
        answer = input(prompt).strip().lower()
        if answer in options:
            return answer
        print(f"❌ Choose from: {list(options)}")

def get_fortune(color, number):
    if random.randint(1, 100) <= 7:
        fortune = random.choice(RARE_FORTUNES)
    else:
        fortune = f"🔮 Number {number} reveals: {NUMBER_MEANINGS[number]}"
    history.append(fortune)
    return fortune
```

## 3. Main Game Loop

```python
def fortune_teller():
    print("\n━━━━━━━━━━━━ 🔮 MYSTIC ORACLE 🔮 ━━━━━━━━━━━━")
    time.sleep(0.5)
    print("✨ I am the ancient seer. I will reveal your future...\n")

    while True:
        color = ask("🎨 Choose a color [yellow, green, blue, red]: ", PREDICTIONS.keys())
        reveal_numbers(color)
        number = int(ask("\n🔢 Choose your number: ", [str(n) for n in PREDICTIONS[color]]))
        mystical_loading()
        print(f"\n📜 Your Divine Fortune: {get_fortune(color, number)}\n")
        if ask("🔄 Another reading? (y/n): ", ["y","n"]) == "n":
            print("\n━━━━━━━━━━━━ 📜 SESSION SUMMARY 📜 ━━━━━━━━━━━━")
            for i, h in enumerate(history, 1):
                print(f"  {i}. {h}")
            print("✨ The stars have spoken... until next time! 🌙\n")
            break
```

## ▶ Run the Oracle

```python
fortune_teller()
```
